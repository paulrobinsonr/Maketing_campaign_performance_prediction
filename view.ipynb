```python
mean_cols = [
    "Duration",
    "Impressions",
    "Engagement_Score"
]

median_cols = [
    "Clicks",
    "Leads",
    "Conversions",
    "Revenue",
    "Acquisition_Cost"
]

df[mean_cols] = df[mean_cols].fillna(df[mean_cols].mean())

df[median_cols] = df[median_cols].fillna(df[median_cols].median())
```

```python
categorical_missing_cols = [
    "Campaign_Type",
    "Target_Audience",
    "Channel_Used",
    "Language",
    "Customer_Segment"
]

df[categorical_missing_cols] = df[categorical_missing_cols].fillna("Unknown")
```

```python
import matplotlib.pyplot as plt

df[numerical_cols].plot(
    kind="box",
    subplots=True,
    layout=(3, 3),
    figsize=(15, 10)
)

plt.tight_layout()
plt.show()
```

```python
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

    print(f"{col}: {outliers} outliers")
```

```python
total_rows = len(df)

for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

    percentage = (outliers / total_rows) * 100

    print(f"{col}: {outliers} outliers ({percentage:.2f}%)")
```
